### K-Means Clustering
- Particular range plot the Ebow method and find the number of qualified clusters.
- kmeans.inertia_ is mathematical property in kmeans, is used to find the least cutting (qualified clusters)
- then using the n_clusters, train and predict the KMeans
- using the predicted category, make supervised dataset (cluster value will be target variable here) and pass it to team who doing the target marketing.

### Find the number of possible clusters using Elbow method

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=5, init="k-means++", random_state=42)

# There is no need the fit and then predict. At the training time itself it will predict
y_kmeans = kmeans.fit_predict(X)
```
